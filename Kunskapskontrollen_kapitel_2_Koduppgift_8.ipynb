```python
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from joblib import dump, load

# Skapa syntetiskt dataset med 20000 observationer och 3 features
X, y = make_regression(n_samples=20000, n_features=3, noise=0.1)

# Träna en linjär regressionsmodell på datan
model = LinearRegression().fit(X, y)

# Spara den tränade modellen till fil
dump(model, "linear_model.joblib")

# Ladda tillbaka modellen från fil
loaded_model = load("linear_model.joblib")

# Använd den laddade modellen för att prediktera de första 5 observationerna
print(loaded_model.predict(X[:5]))
```

##### 8.	Från sklearn importerar vi make_regression som används för att skapa ett syntetiskt dataset för regression med 20 000 observationer och 3 features. Från sklearn importerar vi även LinearRegression som är modellen som tränas på datan. Från joblib importerar vi dump och load som är verktyg för att spara den tränade modellen till filen "linear_model.joblib" och sedan ladda tillbaka den.

##### Den inlästa modellen används slutligen för att prediktera de första 5 raderna, vilket skrivs ut.
